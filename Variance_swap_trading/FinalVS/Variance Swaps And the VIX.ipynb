## What are Variance swaps?

There is a vast amount of material on explaining what variance swaps are and some of the basic mathematics behind it. Here is a breif explanation.
Variance swaps are derivative contracts with a payoff fucntion based of the realised variance over an agreed upon time period. With the buyer agreeing to pay a strike price, $ K$ and the seller paying out the $ v(T) $ the realised variance over that period. Variance swaps dont like usual swaps where cash flows are exchanged over a period, this is a payment of depedning on the volatility over that period. 


```python
import pandas as pd  
import numpy as np
import matplotlib.pyplot as plt
```

The payoff of a variance swap is defined as being  
$$
P(T) = N \times (v(T)- K) 
$$
where N is the notional amount. For variance swaps its quoted in vega notional, where market pratice is 
$$
N = \frac{Vega Notional}{2K}
$$

## Why Variance Swaps?

Why do we choose Variance swaps to trade volatilty. Firstly, convexity, being able to gain non-linear payoffs can generate very large returns espcially in large tail risk situations. This makes variance swaps a good tail risk hedge as they are long 'vol-of-vol'. Shorting Variance swaps allows us to take advantage of the VRP idea mentioned as they recive a market expected volatitliy, and pay the realised volailtity over the agreed upon time period. This allows investors with a method to take advantage of the VRP. 

Above we see the convex nature and also prove it mathematically:
$$
P(T) = N \times (v(T)- K) 
$$
where $ v(T) = \sigma ^2 $
$$
\frac{dP(T)}{d\sigma} = 2N\sigma
$$

$$ 
\frac{d^2P(T)}{d\sigma^2} = 2N > 0 
$$
